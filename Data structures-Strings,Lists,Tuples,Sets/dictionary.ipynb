Dictionary examples

```python
setobj = set()#empty set
dobj1={} #empty dictionary
dobj={"DptId":10,"DptName":"Strategy","DptLoc": "Pune"}
print(dobj)

#Data accessing using keys
print("Dpt Id:",dobj["DptId"])

#update values in existing objects
dobj["DptId"]=20
dobj["DptLoc"]="Mumbai"
print(dobj)

print(dobj.get("DptId"))
for key in dobj.keys():
    print("key name:",key,"Value",dobj.get(key))

keys=dobj.keys()
print(keys)

Values=dobj.values()
print(Values)

print(dobj.items())
for item in dobj.items():
    print(item)
```

```python
#nested dictionary

dobj2={1001:{"Emp Name":"Ganesh","Location":"Pune"},
      1002:{"Emp Name":"Ram","Location":"Mumbai"},
      1003:{"Emp Name":"Dinesh","Location":"Pune"}
        }

print(dobj2[1001])

print(len(dobj2[1001]))
```

```python
#Integration of Types

dobj3={
    1001:["Hello","Hi"],
    1002:("Welcome","Bye"),
    1003:{12,13,14,56}
}

for item in dobj3.items():
    print(item) #converts into tuple ,because it is most safest(cannot be changed)

dobj4=dobj3.copy()
dobj3.clear
print(dobj4)
```
