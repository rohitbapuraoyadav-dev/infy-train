```python
setobj=set()#empty set use set keyword
setobj2={101,"Ganesh",2345,"Pune"} #no need to use keyword,only use it to intialise

print(setobj2)

#update and modify
#setobj2[0]=102 #does not support object asignment

setobj2.add(23452346)
print(setobj2)
setobj2.remove(23452346)
print(setobj2)
setobj2.discard("Mumbai")
print(setobj2)

```

```python
set1={1,2,3,4,5,6}
set2={7,8,9,0,12,6}

print("Uncommon Values:",set2.difference(set1)) #diffrence
print("symmetric diff:",set1-set2)
print("Sym Diff" , set1^set2)# symmetric diff
print(set1.issuperset(set2))

print("intersection/common Values:",set1.intersection(set2))
print("intersection:",set1&set2)
print(set1.issubset(set2))
print(set1.issuperset(set2))
print("Are there common values :",set1.isdisjoint(set2))
print("Union:",set1.union(set2))
print("union:",set1|set2)
print(set1.pop())
#set1.clear()#remove all values
print("length:",len(set1))
```
