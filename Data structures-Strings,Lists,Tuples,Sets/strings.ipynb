```python
str2="welcome to Kaiser Permenante,GCC,Pune"
print("Capitalize:",str2.capitalize())
print("Casefold:",str2.casefold())
if (str2.islower()):
    print("IS Lower Case")
else:
    print("Is Mixed case")

if(str2.isupper()):
    print("is upper case:")
else: 
    print("Is mixed case")

print(str2.swapcase())
print(str2.title()) #important for documentation

if str2.istitle():
    print("Title Case:")
else:
    print("Not Title Case")
```

```python
#python string methods to validate character

str3="w@lco$@ _o _he Jungl@ of !$azon~ "
str4="Welcome123"
if str4.isalnum():
    print("It is alphanumeric")
else:
    print("No alphanumeric values")

if str4.isalpha():
    print("Aplha Values")
else:
    print("No alphabets")

num1="345.989"
if num1.isdecimal():
    print("Decimal Value:")
else:
    print("Not a decimal:")

num1="6764"
if num1.isdigit():
    print("Digit Value:")
else:
    print("Not a Digit Value:")

num1="6764"
if num1.isnumeric():
    print("Numeric Value:")
else:
    print("Not a Numeric Value:")

num1="   "
if num1.isspace():
    print("Space")
else:
    print("Not a spacre ")

num1="the adkjl"
if num1.isprintable():
    print("Print Value")
else:
    print("CanNot a print")

```
