```python
#calculate variance

Population = [5 , 7, 9, 10, 14 ,14]
Mean = sum(Population)/len(Population)
print(round(Mean,3))

sumsqr = 0
for i in range(0, len(Population)):
    diffval = Mean - Population[i]
    sumsqr += diffval**2
variance = sumsqr/len(Population)
print(round(variance,3))
```

```python
#calculate variance

data=[5 ,7 ,9 ,10 ,14 ,15]

sum = 0
for i in data :
    sum += i
    Mean = sum/len(data) # here the placement should be shifted ,otherwise after sum it will go to mean and print it for each value.
print(Mean)

#
sumsqr = 0
for i in data :
    diffval = i - Mean
    sumsqr += pow(diffval , 2)
    variance = sumsqr/len(data)
print(round(variance,3))
```

```python
#skewness
from math import sqrt

data=[85, 88, 92, 94, 96, 98, 100, 100, 100, 100]

sum = 0
for i in data :
    sum += i
Mean = sum/len(data) # here the placement should be shifted ,otherwise after sum it will go to mean and print it for each value.
print(Mean)

Median = (data[4]+data[5])/2
print(Median)

# SD

sumsqr = 0
for i in data :
    diffval = i - Mean
    sumsqr += pow(diffval , 2)
    variance = sumsqr/len(data)
    sd=sqrt(variance)
print(round(sd,3))

#skewness

skw=(3**(Mean-Median))/sd
print("Skewness:",skw)
```
