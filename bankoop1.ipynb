```python
class Bank():
    account_no = 100001 # initiate account no
    def __init__(self ,acc_hol ,IFSC_Code ,bal ) :
        self.acc_no=Bank.account_no
        Bank.account_no += 2
        self.acc_hol= acc_hol
        self.IFSC_Code = IFSC_Code
        self.bal = bal

    def show_bank_details(self):
        return ("Account No :", self.acc_no,"Name :",self.acc_hol ,"Balance :",self.bal ,"IFSC Code :", self.IFSC_Code)

# saving acc under the parent class Bank

class Savings(Bank):
    def __init__(self,acc_hol ,IFSC_Code ,bal ,deposit ,withdraw):
        super().__init__(acc_hol ,IFSC_Code ,bal)

#check min bal

        if deposit<0:
          raise ValueError("Deposit has to be positive")
        
        self.deposit=deposit
        self.bal +=deposit

    #check withdraw condition
        if withdraw>self.bal:
            raise ValueError("Insufficient Balance")
        print("You can withdraw,mention the amount :")
        
        self.withdraw=withdraw
        self.bal -=withdraw

    def show_saving_acc_details(self):
        return("Account No :", self.acc_no,"Name :",self.acc_hol ,"Balance :",self.bal ,
               "IFSC Code :", self.IFSC_Code,
               "Deposit Amount :",self.deposit ,"Withdrawal:", self.withdraw)
    def current_balance():
        return ("Current Balance :", self.bal)

class Current(Bank):
    def __init__(self,acc_hol ,IFSC_Code ,bal ,overdraft ,withdraw ): 
        super().__init__(acc_hol ,IFSC_Code ,bal)
        self.overdraft = overdraft

        #allowed max limit
        max_allowed=self.bal + self.overdraft

        if withdraw > max_allowed :
            raise ValueError("Overdraft limit exceeded:")

        print("Withdrawal  allowed.")
        self.withdraw=withdraw
        self.bal -= withdraw

    def show_current_acc_details(self):
        return ("Account No :", self.acc_no,"Name :",self.acc_hol ,"Balance :",self.bal ,
               "IFSC Code :", self.IFSC_Code,
               "Overdraft:",self.overdraft ,"Withdrawal:", self.withdraw)
```
