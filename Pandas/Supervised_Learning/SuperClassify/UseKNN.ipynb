```python
# import relevant libraries

import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score,r2_score,recall_score,precision_score,accuracy_score
from sklearn.preprocessing import OneHotEncoder,StandardScaler

#Load data

# hdf = pd.read_csv("houseprice_truncated.csv")

cdf = pd.read_csv("Credit.csv")

# Data cleaining,Encoders,Scaler
# ohe = OneHotEncoder()
# neighborhood_encoded = ohe.fit_transform(hdf[['Neighborhood']]).toarray()
# neighborhood_df = pd.DataFrame(neighborhood_encoded, columns=ohe.get_feature_names_out(['Neighborhood']))

#select relevant features and target variable
# X = pd.concat([hdf[['YearBuilt','YrSold','total_area']], neighborhood_df], axis=1)
# y = hdf['SalePrice']


X = cdf[['DebtRatio','age','NoRealEstateLoansOrLines','NoofTime30_59DaysPastDue',
         'NoofTime30_59DaysPastDue', 'NoOfTime60_89DaysPastDue']]

Y = cdf['Credit']

#Train Test Split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3,random_state=42)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)



# Create model class
# model = KNeighborsRegressor(n_neighbors=5) #continuous values

model = KNeighborsClassifier(n_neighbors=5) # we use classifier because credit is a binary value

# Train the model
model.fit(X_train,Y_train)

predictions =  model.predict(X_test)

#Check the accuracy metrics of the model

# r2 = r2_score(Y_test,predictions)
# print(f"R2 Score: {r2}")

print("Recall Score:", recall_score(Y_test, predictions))
print("Precision Score:", precision_score(Y_test, predictions))
print("Accuracy Score:", accuracy_score(Y_test, predictions))
print("F1 Score:", f1_score(Y_test, predictions))

#predict credit worthiness of a customer with the given features.

customer= [[0.30, 45, 1, 0, 0, 0]] # example customer features
customer_scaled = scaler.transform(customer)
prediction = model.predict(customer)
probability = model.predict_proba(customer)  # Get the probability of each class
print(f"Probability of being creditworthy (1): {probability[0][1]:.2f}")
print(f"Predicted Creditworthiness: {prediction[0]}")  # Output will be 0 or 1 indicating creditworthiness

#predict the creditworthiness of a customer with the given features.
# giving real life data to the model for prediction

# location = [1 if col == 'Neighborhood_CollgCr' else 0 for col in neighborhood_df.columns]
# house_features = np.array([[2005, 2020, 2500] + location])
# prediction = model.predict(house_features)
# print(f"Predicted Price: {prediction[0]}")
```
