```python
#import labiraries

import pandas as pd
import numpy as np
from sklearn.naive_bayes import GaussianNB,BernoulliNB,MultinomialNB,CategoricalNB
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
from sklearn.metrics import precision_score,r2_score,f1_score,classification_report
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

# load the data into dataframe

ndf= pd.read_csv('sms_spam_dataset.csv',encoding='ISO-8859-1').iloc[:,0:2]
ndf.describe()
ndf.info()
ndf.head()


#use tfid vectorizer

tfid_vectorizer = TfidfVectorizer(stop_words='english',token_pattern=r'\b[a-zA-Z]+\b') #only keep alphabetic values
tfid_matrix = tfid_vectorizer.fit_transform(ndf['v2'])
tfid_df = pd.DataFrame(tfid_matrix.toarray(),columns=tfid_vectorizer.get_feature_names_out())
tfid_df

#encode the data using label encoder.eg.spam=0 and not spam = 1

x= tfid_matrix
y=ndf['v1'].map({'ham':0,'spam':1}) 

# train test split the data 
X_train,X_test,Y_train,Y_test = train_test_split(x,y ,test_size=0.3,random_state=42)

# create model
model= MultinomialNB()
# train the model

model.fit(X_train,Y_train)
Y_test_pred= model.predict(X_test)

# testingb the model with sample data

new_data= pd.DataFrame([['How are you doing? Did you finish your work?'],['Nice Weather!Lets go for a walk'],['Free! Free! Absoulteely Free!']],columns=['v2'])


new_data_vector=tfid_vectorizer.transform(new_data['v2'])
predictions=model.predict(new_data_vector)
print(predictions)

#metrics of model

print("F1Score:",f1_score(Y_test,Y_test_pred))
print("R2 Score:",r2_score(Y_test,Y_test_pred))
print("Precision Score :",precision_score(Y_test,Y_test_pred))
print("Classification Report:",classification_report(Y_test,Y_test_pred))
```
