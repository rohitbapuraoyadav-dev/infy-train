```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes,load_wine,load_breast_cancer,load_iris
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.metrics import f1_score,r2_score,recall_score,precision_score,accuracy_score,classification_report





# diabetes=datasets(as_frame=True).frame
# diabetes.describe()
# diabetes.info()



diabetes=datasets.load_diabetes()
X= diabetes.data # Features
y = (diabetes.target >140).astype(int)

# create data frame for better visvualisation
diabetes_df= pd.DataFrame(data=X,columns=diabetes.feature_names)
diabetes_df['diabetes']=y
diabetes_df.head()
#find the correlation between diffrent variables 
corr_target =diabetes_df.corr()['diabetes'].sort_values(ascending=False)
print(corr_target)

#load the truncated dataframe
# diabetes_df=diabetes_df.drop(columns=['sex','s3'])
# diabetes_df

#select relevant features and target variable

X = diabetes_df[['age','bmi','bp','s1','s2','s4','s5']]
# Y = diabetes_df['']

##Train Test Split

X_train,X_test,Y_train,Y_test = train_test_split(X,y,test_size=0.3,random_state=42)

# scalar = StandardScaler()

# X_train = scalar.fit_transform(X_train)
# X_train = scalar.transform(X_test)

#create model class

model= KNeighborsClassifier(n_neighbors=5)

print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)


#train the model 

model.fit(X_train,Y_train)
predict=model.predict(X_test)

#check the accuracy
print("F1 Score:", f1_score(Y_test, predict))
print("Classification report:",classification_report(Y_test,predict))


```
