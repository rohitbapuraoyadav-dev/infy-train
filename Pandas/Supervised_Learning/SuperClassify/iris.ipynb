```python
#import the libraries 
import numpy as pd
import pandas as pd
from sklearn.svm import SVC
from sklearn import datasets
from sklearn.metrics import f1_score,r2_score,recall_score,precision_score,accuracy_score,classification_report
from sklearn.model_selection import train_test_split




#import dataset
iris=datasets.load_iris()


#choose features and target values
X = iris.data
Y = iris.target

#train test split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y ,test_size=0.3,random_state=42)
#create model instance
model = SVC(kernel='poly')

#train the model
model.fit(X_train,Y_train)

#precitions and metrics
predict=model.predict(X_test)

print('\n Classification Report:',classification_report(Y_test,predict))

```

```python
#import the libraries 
import numpy as pd
import pandas as pd
from sklearn.svm import SVC
from sklearn import datasets
from sklearn.metrics import f1_score,r2_score,recall_score,precision_score,accuracy_score,classification_report
from sklearn.model_selection import train_test_split




#import dataset
cdf=pd.read_csv("Credit.csv")


#choose features and target values
X = cdf[['DebtRatio','age','NoRealEstateLoansOrLines','NoofTime30_59DaysPastDue',
         'NoofTime30_59DaysPastDue', 'NoOfTime60_89DaysPastDue']]

Y = cdf['Credit']
#train test split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y ,test_size=0.3,random_state=42)
#create model instance
model = SVC(kernel='poly')

#train the model
model.fit(X_train,Y_train)

#precitions and metrics
predict=model.predict(X_test)

print('\n Classification Report:',classification_report(Y_test,predict))

```
