```python
import pandas as pd
import numpy as nd
from sklearn.ensemble import VotingClassifier,VotingRegressor
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score,classification_report,r2_score,accuracy_score
from sklearn.model_selection import train_test_split
from sklearn import datasets

#load dataset
cancer=datasets.load_breast_cancer()
cancerdf=pd.DataFrame(cancer.data,columns=cancer.feature_names)
cancerdf['cancer']= cancer.target
X = cancer.data
Y = cancer.target

# corr_target =cancerdf.corr()['cancer'].sort_values(ascending=False)
# print(corr_target)
print(cancerdf.columns)

#train test split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y ,test_size=0.3,random_state=42)

#choose models and create Ensemble model:

treemodel= DecisionTreeClassifier(max_depth=5,max_leaf_nodes=10)
svcmodel = SVC(kernel='rbf')
logmodel = LogisticRegression(max_iter=10000)

classifiers= [('dt',treemodel),('logr',logmodel),('svc',svcmodel)]

en_model = VotingClassifier(estimators= classifiers,voting='hard')

en_model.fit(X_train,Y_train)

#predict and metrics

prediction = en_model.predict(X_test)
accuracy = accuracy_score(Y_test,prediction)

print("Accuracy :",accuracy)
```
