```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes,load_wine,load_breast_cancer,load_iris
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.svm import SVC
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import f1_score,r2_score,recall_score,precision_score,accuracy_score,classification_report,confusion_matrix

#load dataset
hdf = pd.read_csv('heart_disease.csv')

#data overview
# print(hdf.describe())
print(hdf.head())

#correlation

#distribution of target variable
sns.countplot(x='target',data=hdf ,palette='Set2')
plt.title("Distribution of Heart Diseases")
plt.show()

X = hdf.drop ('target',axis=1)
Y = hdf['target']

#pairplot for selection


#model creation

model=GradientBoostingClassifier(loss='log_loss',n_estimators=100,learning_rate=0.05,max_depth=5,subsample=0.1)

#test train split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y ,test_size=0.3,random_state=42)

#train

model.fit(X_train,Y_train)

#predict

predictions=model.predict(X_test)

#metrics

print("Accuracy:",accuracy_score(Y_test,predictions))

print("Classification Report:",classification_report(Y_test,predictions))

```
