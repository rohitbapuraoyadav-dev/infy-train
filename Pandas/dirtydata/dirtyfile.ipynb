```python
import pandas as pd

filedf=pd.read_csv("my_file.csv")
filedf

print("Count of empty values in All Time Rank:",filedf["All Time Peak"].isna().sum())
filedf
filedf["Actual gross"]=filedf["Actual gross"].str.replace("$","")
filedf["Actual gross"]=filedf["Actual gross"].str.replace(",","")
filedf["Actual gross"]=filedf["Actual gross"].str.replace("[b]","")
filedf["Actual gross"]=filedf["Actual gross"].str.replace("[e]","")


filedf["Year(s)"] = filedf["Year(s)"].str.split("–").str[0]
filedf

filedf["Average gross"]=filedf["Average gross"].str.replace("$","")
filedf["Average gross"]=filedf["Average gross"].str.replace(",","")

filedf

filedf["Actual gross(in 2022 dollars)"]=filedf["Actual gross(in 2022 dollars)"].str.replace("$","")
filedf["Actual gross(in 2022 dollars)"]=filedf["Actual gross(in 2022 dollars)"].str.replace(",","")
filedf

#converting string values into int
filedf["Actual gross"]=pd.to_numeric(filedf["Actual gross"],errors="coerce")
filedf["Actual gross(in 2022 dollars)"]=pd.to_numeric(filedf["Actual gross(in 2022 dollars)"],errors="coerce")
filedf["Average gross"]=pd.to_numeric(filedf["Average gross"],errors="coerce")
filedf

filedf.to_csv("cleanfile.csv")
filedf.to_excel("cleanfile.xlsx")
```

```python
#calculate quartile
import pandas as pd

filedf=pd.read_csv("my_file.csv")
filedf["Actual gross"] = pd.to_numeric(filedf["Actual gross"], errors="coerce")
Q1=filedf["Actual gross"].quantile(0.25)
Q3=filedf["Actual gross"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1-5*IQR
upper=Q3+1.5*IQR
outliers=filedf[(filedf<lower)|(filedf>upper)]
print(outliers)
```
