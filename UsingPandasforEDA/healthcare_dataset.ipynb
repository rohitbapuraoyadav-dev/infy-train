```python
import pandas as pd

df=pd.read_csv("healthcare_dataset.csv")

#calculate outliers using IQR method

Q1 = df["Billing Amount"].quantile(0.25)
Q3 = df["Billing Amount"].quantile(0.75)

IQR = Q3 - Q1
lower = Q1 - 0.60 * IQR
upper = Q3 + 0.60 * IQR

outliers = df[(df["Billing Amount"]<lower) | (df["Billing Amount"]>upper)]
print("Outliers in Billing Amount:", outliers)
```

```python
#calculating z score (also use manually calculated z score formula) to identify outliers)


import pandas as pd

from scipy import stats

df=pd.read_csv("healthcare_dataset.csv")

z_scores = stats.zscore(df["Billing Amount"])

outliers = df[abs(z_scores) > 1.9]
print("Outliers in Billing Amount:", outliers)
```

Q7. Multivariate Outlier Detection
Using:
	• Age
	• Billing Amount
	• Length of Stay
Highlight records that appear normal individually but anomalous collectively.

```python
import pandas as pd

df=pd.read_csv("healthcare_dataset.csv")

df["Date of Admission"]=pd.to_datetime(df["Date of Admission"],errors="coerce")
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"],errors="coerce")

df["len_of_stay"]=(df["Discharge Date"] - df["Date of Admission"]).dt.days
df=df[df["len_of_stay"]>=0]
print((df["len_of_stay"]<0).sum())

print(df[["Date of Admission", "Discharge Date", "len_of_stay"]].head())

Q1=df[["Date of Admission","len_of_stay","Billing Amount"]].quantile(0.25)
Q3=df[["Date of Admission","len_of_stay","Billing Amount"]].quantile(0.75)
IQR=Q3-Q1
upper=Q3+1*IQR
lower=Q1-1*IQR

outliers=df[(df[["Date of Admission","len_of_stay","Billing Amount"]]>upper)|(df[["Date of Admission","len_of_stay","Billing Amount"]]<lower)]
print("Outliers in Date of Admission, len_of_stay, Billing Amount:", outliers.count())
```
