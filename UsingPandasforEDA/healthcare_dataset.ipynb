```python
import pandas as pd

df= pd.read_csv("healthcare_dataset.csv")

#• Identify inconsistent capitalization, prefixes (Dr., MD, DDS), and formatting issues.

# df.describe()
# df.head()
# df.tail()
# df.info()

df["Name"]=df["Name"].str.title()
# print("Corrected Names",df["Name"])


df["Hospital"]=(df["Hospital"].str.title().str.rstrip(",").str.strip().str.replace(r"\bAnd\b","and",regex=True))
# print("Corrected Names of Hospitals:",df["Hospital"])


df["Date of Admission"]=pd.to_datetime(df["Date of Admission"],errors="coerce")
# print("Corrected Dates :",df["Date of Admission"])

df["Discharge Date"]=pd.to_datetime(df["Discharge Date"],errors="coerce")
# print("Corrected Dates:",df["Date of Admission"])


df["Doctor"]=df["Doctor"].str.title()
# print("Corrected Names of Doctors:",df["Doctor"])

df["Gender"]=df["Gender"].map({"Male": 0,"Female": 1})


df = df[["Age","Gender","Medical Condition","Admission Type","Medication","Test Results"]]

print("Correlation of Test Results with Admission Type:\n",
      pd.crosstab(df["Admission Type"], df["Test Results"], normalize="index").round(2) * 100)

pd.crosstab([df["Gender"],df["Medical Condition"],df["Admission Type"],df["Medication"]],df["Test Results"],normalize="index").round(2) * 100
```
