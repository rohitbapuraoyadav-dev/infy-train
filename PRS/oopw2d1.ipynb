```python
class Employee():
 
 emp_counter = 1000 #class variable for auto-generating IDs

 def __init__(self,empname,empsalary):
     Employee.emp_counter += 1
     self.empid = Employee.emp_counter      
     self.empname = empname   
     if empsalary<=0:
        raise ValueError("incorrect value")
     else: 
        self.empsalary = empsalary


 def showdetails(self):   
     print("Employee ID:", self.empid)       
     print("Employee Name:", self.empname)     
     print("Employee Salary:", self.empsalary)


class Manager(Employee):

  def __init__(self, empname, empsalary, perks):     
      super().__init__(empname, empsalary) #super is used to adopt from 3 parameters of Employee-id,name,salary
      self.perks = perks 

  def showdetails(self):      
      print("Manager Details")        
      print("Perks:", self.perks)

class Salesman(Employee):

  def __init__(self, empname, empsalary, commission):     
      super().__init__(empname, empsalary) #super is used 
      if commission<0:
         raise ValueError("commission cannot be -ve")
      else :
        self.commission = commission

  def showdetails(self):      
      print("Salesman Details")        
      print("Commission:", self.commission)

empname = input("Enter valid name :")
empsalary = float("Enter valid salary :")

```

```python
# object 
m1 = Manager("Alice",0,5000)
s1 = Salesman("Jhon",40000,-3000)
m1.showdetails()
s1.showdetails()
```
